```python
import tiktoken
```

```python
# encodings specify how text is converted int tokens

encoding = tiktoken.encoding_for_model("gpt-4o-mini")

# .encode() method converts a text string into a list of token integers.
encoding.encode("what day is it?")
```

```python
encoding.decode([13347, 2163, 382, 480, 30])
```
