```python
import requests


def generate_embeddings(content):
    url = "http://127.0.0.1:8080/embedding"
    headers = {"Content-Type": "application/json"}
    data = {"content": content}
    response = requests.post(url, headers=headers, json=data)
    print(response.json())
    return response.json()[0]["embedding"]


generate_embeddings("The president is probably in Dubai right now, but he will be back in the Kenya soon. Who knows what he is doing there, but he is probably having a good time.")
```
